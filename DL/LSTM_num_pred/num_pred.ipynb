```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

- LSTM이란
- 입력(현재값) + 내부 상태 (과거의 상태를 요약한 상태) 이 두 데이터를 잘 조합해서 다음 출력을 만드는 것 
- Dense는 숫자나 데이터를 섞어서 (가중치) 새로운 숫자로 바꾼다
- 입력 벡터 -> 선형 변환 (두 변수의 변화량에 따라)-> 활성화(얼마의 값을 줄지 결정)
- 단어 맞추기도 가능하지만 
- 단어의 경우 숫자를 읽을 수 있는 컴퓨터에 맞춰서 변환의 과정이 필요하다
- 숫자 시퀀스는 LSTM  / 

```python
data = np.array([i for i in range(1, 21)])
timestemps =3 
```

- range(1, 21)
    - 1부터 21까지 1씩 증가
- 값을 3개씩 묶어서 본다

```python
x_data = [] # 기존 기록을 저장할 변수
y_data = [] # 새로운 출력값을 저장할 변수 
for i in range(len(data) - timestemps):
    x_data.append([data[i : i +timestemps]])
    y_data.append([data[i+timestemps]])

x_data = np.array(x_data).reshape(-1, timestemps, 1)
y_data = np.array(y_data)

for i in range(len(x_data)):
    print(f"x: {x_data[i].reshape(-1)} -> y : {y_data[i]}")
```

```python
model = Sequential([
    LSTM(10, input_shape=(timestemps, 1)),
    Dense(1)
])
```

```python
model.compile(optimizer='adam', loss = 'mse')
```

```python
model.fit(x_data, y_data, epochs=400, verbose=0)
```

```python
x = input("3개의 숫자를 입력하세요").split()
x = list(map(int, x))

test_input = np.array(x).reshape(1, timestemps, 1)

pred_velue = model.predict(test_input)
print(pred_velue)

print("입력 데이터", test_input.flatten())
print("예측값" , pred_velue[0][0])
```

```python
# 4. 예측 테스트
test_input = np.array([18, 19, 20]).reshape(1, timestemps, 1)  # 입력 예시
predicted_value = model.predict(test_input)
```
